```python
texts = [f"Sample sentence number {i} about cloud computing and vector search at scale." for i in range(2000)]
print(len(texts))
```

```python
from sentence_transformers import SentenceTransformer

def measure(device, batch_size):
    model = SentenceTransformer("all-MiniLM-L6-v2", device=device)
    model.encode(texts[:64], batch_size=batch_size)   # warm-up, not timed
    start = time.time()
    model.encode(texts, batch_size=batch_size)
    seconds = time.time() - start
    return len(texts) / seconds
```

```python
for device in ["cpu", "cuda"]:
    for bs in [1, 8, 32, 128]:
        speed = measure(device, bs)
        print(device, "batch size", bs, "->", round(speed, 1), "sentences/sec")
```
